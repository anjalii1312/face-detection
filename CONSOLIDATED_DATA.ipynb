```python
import cv2
import os
import pickle#file format where you can store files like zip
import numpy as np
```

```python
data_dir = os.path.join(os.getcwd(),'data')#u get the access to the files in the directory
img_dir = os.path.join(os.getcwd(),'images')
```

```python
print(os.getcwd()) #getcwd - get current working directory
```

```python
image_data = []
labels = []
```

```python
for i in os.listdir(img_dir):
    print(i)
```

```python
#preprocessing steps
for i in os.listdir(img_dir):
    image = cv2.imread(os.path.join(img_dir,i))
    image = cv2.resize(image,(100,100))
    image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)#to train image fast
    image_data.append(image)
    labels.append(str(i).split("_")[0])#to just take the name before '_'
```

```python
image_data = np.array(image_data)    
labels = np.array(labels) 
```

```python
print(image_data)
```

```python
print(labels)
```

```python
import matplotlib.pyplot as plt
plt.imshow(image_data[50],cmap="gray")
plt.show()
```

```python
with open(os.path.join(data_dir,"images.p"),'wb') as f:#pickle file extension .p
    pickle.dump(image_data,f) #dump - store
```

```python
with open(os.path.join(data_dir,"labels.p"),'wb') as f:
    pickle.dump(labels,f)
```
